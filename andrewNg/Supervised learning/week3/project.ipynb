    The Engine: It multiplies those numbers by the optimized weights it learned during training and squashes the result through a Sigmoid curve.

    The Probability: It outputs a mathematical percentage representing its confidence (e.g., "There is an 82% chance this user will purchase").

    The Output: It applies a hard threshold (usually 50%). If the probability is ≥0.5, it returns a 1 (Will Buy). If it is <0.5, it returns a 0 (Will Not Buy).
