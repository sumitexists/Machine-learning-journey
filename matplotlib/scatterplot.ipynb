### Learning the Scatter Plots

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
print(plt.style.available)
plt.xkcd()
```

```python
data = pd.read_csv('2019-05-31-data.csv')
view_count = data['view_count']
likes = data['likes']
ratio = data['ratio']
```

```python
plt.scatter(view_count,likes,edgecolors='black', alpha=0.75, linewidths=1, c=ratio, cmap='summer')
cbar = plt.colorbar()
cbar.set_label('Like/Dislike Ratio')
plt.xscale('log')
plt.yscale('log')
plt.title('Scatterplot of Trending YouTube Videos')
plt.xlabel('Total View Count')
plt.ylabel('Total Likes')
plt.tight_layout()
plt.savefig('scatterplot.png', dpi=300)
plt.show()
```
