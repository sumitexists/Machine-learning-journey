## Learning subplots

```python
import pandas as pd
from matplotlib import pyplot as plt
plt.style.use('seaborn-v0_8-darkgrid')
```

```python
data = pd.read_csv('data.csv')
ages = data['Age']
dev_salary = data['All_Devs']
py_salary = data['Python']
js_salary = data['JavaScript']
```

```python
fig , (ax1, ax2) = plt.subplots(2,1, sharex=True)
ax2.plot(ages, py_salary, label='Python')
ax2.plot(ages, js_salary, label='JavaScript')
ax1.plot(ages, dev_salary, label='All Devs', linestyle='--', color='#444444')
ax1.legend()
ax1.set_title('Median Salary (USD) by Age')
ax1.set_ylabel('Median Salary (USD)')
ax2.legend()
ax2.set_xlabel('Ages')
ax2.set_ylabel('Median Salary (USD)')
plt.tight_layout()
plt.show()
```
